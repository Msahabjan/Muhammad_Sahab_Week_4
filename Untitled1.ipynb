```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import classification_report, roc_auc_score

# 1. Kaggle Direct Download URL / Public Link se Data Load karein
# (Agar kaggle dataset local upload kar rahe ho toh path update kar lena)
url = "https://storage.googleapis.com/download.tensorflow.org/data/creditcard.csv"
df = pd.read_csv(url)

print("Dataset Shape:", df.shape)

# 2. Features aur Target alag karein
X = df.drop(columns=['Class']).values
y = df['Class'].values.reshape(-1, 1)

# 3. Data Scaling & Train-Test Split
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.2, random_state=42, stratify=y
)

print(f"Train Shape: {X_train.shape}, Test Shape: {X_test.shape}")
```

```python
import numpy as np
import time

# 1. NumPy Neural Network Architecture
class NeuralNetworkScratch:
    def __init__(self, input_dim, hidden_dim, output_dim, learning_rate=0.05):
        self.lr = learning_rate
        # He Normal Initialization for ReLU
        self.W1 = np.random.randn(input_dim, hidden_dim) * np.sqrt(2.0 / input_dim)
        self.b1 = np.zeros((1, hidden_dim))
        # Xavier Initialization for Sigmoid
        self.W2 = np.random.randn(hidden_dim, output_dim) * np.sqrt(1.0 / hidden_dim)
        self.b2 = np.zeros((1, output_dim))

    def _relu(self, Z):
        return np.maximum(0, Z)

    def _relu_derivative(self, Z):
        return (Z > 0).astype(float)

    def _sigmoid(self, Z):
        return 1 / (1 + np.exp(-np.clip(Z, -500, 500)))

    def forward(self, X):
        self.Z1 = np.dot(X, self.W1) + self.b1
        self.A1 = self._relu(self.Z1)
        self.Z2 = np.dot(self.A1, self.W2) + self.b2
        self.A2 = self._sigmoid(self.Z2)
        return self.A2

    def compute_loss(self, y_true, y_pred):
        epsilon = 1e-15
        y_pred = np.clip(y_pred, epsilon, 1 - epsilon)
        loss = -np.mean(y_true * np.log(y_pred) + (1 - y_true) * np.log(1 - y_pred))
        return loss

    def backward(self, X, y_true):
        m = X.shape[0]

        # Output layer gradient
        dZ2 = self.A2 - y_true
        dW2 = (1 / m) * np.dot(self.A1.T, dZ2)
        db2 = (1 / m) * np.sum(dZ2, axis=0, keepdims=True)

        # Hidden layer gradient
        dA1 = np.dot(dZ2, self.W2.T)
        dZ1 = dA1 * self._relu_derivative(self.Z1)
        dW1 = (1 / m) * np.dot(X.T, dZ1)
        db1 = (1 / m) * np.sum(dZ1, axis=0, keepdims=True)

        # Gradient Descent Updates
        self.W2 -= self.lr * dW2
        self.b2 -= self.lr * db2
        self.W1 -= self.lr * dW1
        self.b1 -= self.lr * db1

# 2. Training NumPy Model
input_dim = X_train.shape[1]
hidden_dim = 64
output_dim = 1
batch_size = 2048
epochs = 10

nn_scratch = NeuralNetworkScratch(input_dim, hidden_dim, output_dim, learning_rate=0.05)

print("--- NumPy Model Training Started ---")
numpy_start_time = time.time()

for epoch in range(epochs):
    # Shuffle dataset
    indices = np.arange(X_train.shape[0])
    np.random.shuffle(indices)
    X_shuffled = X_train[indices]
    y_shuffled = y_train[indices]

    for i in range(0, X_train.shape[0], batch_size):
        X_batch = X_shuffled[i:i+batch_size]
        y_batch = y_shuffled[i:i+batch_size]

        y_pred = nn_scratch.forward(X_batch)
        nn_scratch.backward(X_batch, y_batch)

    full_pred = nn_scratch.forward(X_train)
    loss = nn_scratch.compute_loss(y_train, full_pred)
    print(f"Epoch {epoch+1}/{epochs} | Loss: {loss:.4f}")

numpy_total_time = time.time() - numpy_start_time
print(f"NumPy Model Training Completed in {numpy_total_time:.2f} seconds!")
```

```python
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import TensorDataset, DataLoader

# 1. Tensors & DataLoader Setup
X_train_t = torch.tensor(X_train, dtype=torch.float32)
y_train_t = torch.tensor(y_train, dtype=torch.float32)
X_test_t = torch.tensor(X_test, dtype=torch.float32)

train_dataset = TensorDataset(X_train_t, y_train_t)
train_loader = DataLoader(train_dataset, batch_size=2048, shuffle=True)

# 2. PyTorch Model Architecture
class PyTorchNN(nn.Module):
    def __init__(self, input_dim, hidden_dim, output_dim):
        super(PyTorchNN, self).__init__()
        self.fc1 = nn.Linear(input_dim, hidden_dim)
        self.relu = nn.ReLU()
        self.fc2 = nn.Linear(hidden_dim, output_dim)
        self.sigmoid = nn.Sigmoid()

    def forward(self, x):
        x = self.relu(self.fc1(x))
        x = self.sigmoid(self.fc2(x))
        return x

pytorch_model = PyTorchNN(input_dim=X_train.shape[1], hidden_dim=64, output_dim=1)
criterion = nn.BCELoss()
optimizer = optim.SGD(pytorch_model.parameters(), lr=0.05)

# 3. Training Loop
print("--- PyTorch Model Training Started ---")
pytorch_start_time = time.time()

for epoch in range(epochs):
    for X_batch, y_batch in train_loader:
        optimizer.zero_grad()
        outputs = pytorch_model(X_batch)
        loss = criterion(outputs, y_batch)
        loss.backward()
        optimizer.step()

    print(f"Epoch {epoch+1}/{epochs} | Loss: {loss.item():.4f}")

pytorch_total_time = time.time() - pytorch_start_time
print(f"PyTorch Model Training Completed in {pytorch_total_time:.2f} seconds!")
```

```python
import matplotlib.pyplot as plt
from sklearn.metrics import roc_auc_score, classification_report

# --- Experiment 1: Deeper Architecture (2 Hidden Layers) ---
class DeeperPyTorchNN(nn.Module):
    def __init__(self, input_dim):
        super(DeeperPyTorchNN, self).__init__()
        self.network = nn.Sequential(
            nn.Linear(input_dim, 64),
            nn.ReLU(),
            nn.Linear(64, 32),
            nn.ReLU(),
            nn.Linear(32, 1),
            nn.Sigmoid()
        )
    def forward(self, x):
        return self.network(x)

exp1_model = DeeperPyTorchNN(X_train.shape[1])
exp1_optimizer = optim.SGD(exp1_model.parameters(), lr=0.05)

for epoch in range(epochs):
    for X_batch, y_batch in train_loader:
        exp1_optimizer.zero_grad()
        outputs = exp1_model(X_batch)
        loss = criterion(outputs, y_batch)
        loss.backward()
        exp1_optimizer.step()

# --- Experiment 2: Optimizer Switch (SGD -> Adam) ---
exp2_model = PyTorchNN(input_dim=X_train.shape[1], hidden_dim=64, output_dim=1)
exp2_optimizer = optim.Adam(exp2_model.parameters(), lr=0.001)

for epoch in range(epochs):
    for X_batch, y_batch in train_loader:
        exp2_optimizer.zero_grad()
        outputs = exp2_model(X_batch)
        loss = criterion(outputs, y_batch)
        loss.backward()
        exp2_optimizer.step()

# --- Evaluation & ROC-AUC Calculation ---
with torch.no_grad():
    y_pred_scratch = nn_scratch.forward(X_test)
    y_pred_pytorch = pytorch_model(X_test_t).numpy()
    y_pred_exp1 = exp1_model(X_test_t).numpy()
    y_pred_exp2 = exp2_model(X_test_t).numpy()

auc_scratch = roc_auc_score(y_test, y_pred_scratch)
auc_pytorch = roc_auc_score(y_test, y_pred_pytorch)
auc_exp1 = roc_auc_score(y_test, y_pred_exp1)
auc_exp2 = roc_auc_score(y_test, y_pred_exp2)

print("\n================ COMPARISON RESULTS ================")
print(f"1. NumPy (Scratch SGD):    ROC-AUC = {auc_scratch:.4f} | Time = {numpy_total_time:.2f}s")
print(f"2. PyTorch (Baseline SGD): ROC-AUC = {auc_pytorch:.4f} | Time = {pytorch_total_time:.2f}s")
print(f"3. Exp 1 (Deeper Net SGD): ROC-AUC = {auc_exp1:.4f}")
print(f"4. Exp 2 (PyTorch Adam):   ROC-AUC = {auc_exp2:.4f}")
print("====================================================")
```
